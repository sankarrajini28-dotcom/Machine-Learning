```python
import numpy as np
import pandas as pd
```

```python
# Importing dataset
st=pd.read_csv("student_exam_data_v1.csv")
st
```

```python
# Check datatypes
st.info()
```

```python
# Check null values
st.isnull().sum()
```

```python
x=st["study_hours"].value_counts()
x
```

```python
y=st["exam_score"].value_counts()
y
```

```python
#Feature engineering
x=st[["study_hours"]] # Feature
y=st[["exam_score"]]  # Target
```

```python
# Dataset split into train and test data
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.7,random_state=42)
```

```python
# Linear regression
from sklearn.linear_model import LinearRegression
l=LinearRegression()
l.fit(x_train,y_train)
```

```python
y_prediction=l.predict(x_test)
y_prediction
```

```python
# Actual test value
y_test
```

```python
# Accuracy of prediction
from sklearn.metrics import r2_score
r2_score(y_test,y_prediction)
```
