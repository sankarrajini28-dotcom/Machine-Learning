```python
import pandas as pd
df = pd.read_csv("/content/Wholesale customers data.csv")
df
```

```python
from statsmodels.tsa.arima.model import ARIMA
# Define 'data' using an existing numerical column from df, since 'Close' column does not exist.
# For example, using 'Fresh' column.
data = df['Fresh']

model = ARIMA(data, order=(5,1,0))

model_fit = model.fit()

forecast = model_fit.forecast(steps=5)
```

```python
print("Next 5 Days Forecast:")
print("Day 1:", forecast.iloc[0])
print("Day 2:", forecast.iloc[1])
print("Day 3:", forecast.iloc[2])
print("Day 4:", forecast.iloc[3])
print("Day 5:", forecast.iloc[4])
```
