```python
from scipy.optimize import quadratic_assignment
import numpy as np
n = 20
p = 0.7
q = 1 - p
mean = n * p
variance = n * p * q
std =  np.sqrt(variance)

print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", round(std, 2))
```

```python
from scipy.stats import binom
n =10
p = 0.1
exactly_2 = binom.pmf(2,n,p)
none = binom.pmf(0,n,p)
at_least_1 = 1 - none

print("Exactly 2 :", round(exactly_2, 4))
print("No defective:", round(none, 4))
print("At least 1:", round(at_least_1, 4))
```
