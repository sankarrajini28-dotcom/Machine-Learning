```python
import numpy as np
import pandas as pd
```

```python
s=pd.read_csv("Salary_dataset.csv")
s
```

```python
s=s.drop("Unnamed: 0",axis=1)
s
```

```python
#check datatypes
s.info()
```

```python
#To check null values
s.isnull().sum()
```

```python
#Linear regression
s["YearsExperience"].value_counts()
```

```python
s["Salary"].value_counts()
```

```python
#Feature engineering
x=s[["YearsExperience"]]  # Feature
y=s[["Salary"]]           # Target
```

```python
# Dataset split into train data and test data
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.7,random_state=42)
```

```python
#Linear regression
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_prediction=lr.predict(x_test)
y_prediction
```

```python
y_test
```

```python
#Prediction in manually using 3.5 years of experience
lr.predict(np.array([[3.5]]))
```

```python
#Finding the accuracy of prediction with actual test value
from sklearn.metrics import r2_score
r2_score(y_test,y_prediction)
```
