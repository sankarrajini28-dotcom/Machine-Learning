```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import matplotlib.pyplot as plt
```

```python
y = df["Salary"]
```

```python
X = df.drop("Salary", axis=1)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

```python
model = LinearRegression()
```

```python
import numpy as np

# Impute NaNs in X_train with the mean of each column
for i in range(X_train.shape[1]):
    col_mean = np.nanmean(X_train[:, i])
    if np.isnan(col_mean):
        # If the entire column is NaN (should not happen if there are valid values somewhere),
        # fill with 0 or a more appropriate default.
        X_train[:, i] = 0
    else:
        X_train[np.isnan(X_train[:, i]), i] = col_mean

model.fit(X_train, y_train)
```

```python
print("Intercept")

print(model.intercept_)

print()

print("Coefficients")

print(model.coef_)
```

```python
import numpy as np

# Impute NaNs in X_test with the mean of each column
for i in range(X_test.shape[1]):
    col_mean = np.nanmean(X_test[:, i])
    if np.isnan(col_mean):
        # If the entire column is NaN, fill with 0 or a more appropriate default.
        X_test[:, i] = 0
    else:
        X_test[np.isnan(X_test[:, i]), i] = col_mean

predictions = model.predict(X_test)
```

```python
results = pd.DataFrame({

"Actual Salary":y_test,

"Predicted Salary":predictions

})

results.head(10)
```

```python
mae = mean_absolute_error(
    y_test,
    predictions
)

print(mae)
```

```python
mse = mean_squared_error(
    y_test,
    predictions
)

print(mse)
```

```python
rmse = np.sqrt(mse)

print(rmse)
```

```python
r2 = r2_score(
    y_test,
    predictions
)

print(r2)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    y_test,
    predictions
)

plt.xlabel("Actual Salary")

plt.ylabel("Predicted Salary")

plt.title("Actual vs Predicted Salary")

plt.show()
```

```python
residuals = y_test - predictions

plt.figure(figsize=(8,6))

plt.scatter(
    predictions,
    residuals
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Predicted Salary")

plt.ylabel("Residual")

plt.title("Residual Plot")

plt.show()
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Create a pipeline with imputation, scaling, and the linear regression model
pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')), # Impute NaNs with mean
    ('scaler', StandardScaler()),                 # Scale the features
    ('regressor', model)                          # Your LinearRegression model
])

scores = cross_val_score(
    pipeline, # Use the pipeline as the estimator
    X,        # Pass the original X (DataFrame with NaNs)
    y,
    cv=5,
    scoring="r2"
)

print(scores)
print(scores.mean())
```
